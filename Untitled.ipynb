Question 1: What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.

Answer:Deep Learning is a subset of machine learning that uses artificial neural networks with multiple layers to learn patterns from large amounts of data. It is inspired by the way the human brain works.

Evolution of Deep Learning:

Early stage: Simple algorithms like linear regression and decision trees were used.

Machine Learning era: Models improved but required manual feature selection.

Deep Learning era: With more data and powerful computers (GPUs), neural networks with many layers (deep networks) became popular and powerful.

Difference from Traditional Machine Learning:

| Deep Learning                                  | Traditional Machine Learning       |
| ---------------------------------------------- | ---------------------------------- |
| Automatically learns features                  | Requires manual feature selection  |
| Works well with large datasets                 | Works better with smaller datasets |
| Uses deep neural networks                      | Uses simpler models                |
| High accuracy in complex tasks (image, speech) | Limited in handling complex data   |

In short: Deep learning is more advanced and powerful than traditional machine learning, especially for complex problems like image recognition and natural language processing.

Question 2: Explain the basic architecture and functioning of a Perceptron. What are its limitations?

Answer:A Perceptron is the simplest type of artificial neural network used for binary classification (output is 0 or 1).

Basic Architecture:

A perceptron has the following components:

Input layer: Takes input values (x₁, x₂, …)

Weights: Each input has a weight (w₁, w₂, …)

Bias (b): A constant value added to improve flexibility

Summation: Calculates weighted sum → (w₁x₁ + w₂x₂ + … + b)

Activation function: Usually a step function to produce output (0 or 1)

Functioning:

Inputs are multiplied by their respective weights

All values are added along with bias

The result is passed through the activation function

Output is generated:

If value > 0 → Output = 1

Else → Output = 0

Limitations of Perceptron:

Can solve only linearly separable problems

Cannot solve problems like XOR

Uses only a simple step function (not flexible)

Limited learning capability (single layer only)

In short: A perceptron is a basic linear classifier, but it cannot handle complex or non-linear problems.

Question 3: Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.

Answer:
Purpose of Activation Function:

An activation function is used in neural networks to decide whether a neuron should be activated or not. It adds non-linearity to the model, which helps the network learn complex patterns.

Without activation functions, a neural network would behave like a simple linear model and cannot solve complex problems.

Comparison of Sigmoid, ReLU, and Tanh:

| Function    | Range   | Formula (Simple)      | Advantages                                | Disadvantages                    |
| ----------- | ------- | --------------------- | ----------------------------------------- | -------------------------------- |
| **Sigmoid** | 0 to 1  | 1 / (1 + e⁻ˣ)         | Good for binary output                    | Vanishing gradient problem, slow |
| **ReLU**    | 0 to ∞  | max(0, x)             | Fast, widely used                         | Can cause “dead neurons”         |
| **Tanh**    | -1 to 1 | (eˣ - e⁻ˣ)/(eˣ + e⁻ˣ) | Zero-centered output, better than sigmoid | Still has vanishing gradient     |

In Short:

Sigmoid: Used in output layer for binary classification

ReLU: Most commonly used in hidden layers

Tanh: Similar to sigmoid but better centered

Question 4: What is the difference between Loss function and Cost function in neural networks? Provide examples.

Answer:
Loss Function:

A loss function measures the error for a single data point. It shows how far the predicted value is from the actual value.

Example:

For one prediction:
Actual = 5, Predicted = 4 → Loss = (5 − 4)² = 1

Cost Function:

A cost function is the average of all loss values over the entire dataset. It represents the overall performance of the model.

Example:

If losses are: 1, 2, 3

Cost = (1 + 2 + 3) / 3 = 2

Key Differences:

| Loss Function                 | Cost Function                      |
| ----------------------------- | ---------------------------------- |
| Calculated for one data point | Calculated for entire dataset      |
| Measures individual error     | Measures overall error             |
| Used during training steps    | Used to evaluate model performance |

Common Examples:

Mean Squared Error (MSE)

Binary Cross-Entropy

Categorical Cross-Entropy

In short:

Loss function = error for one example

Cost function = average error for all examples


Question 5: What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.

Answer:Role of Optimizers:

Optimizers are algorithms used to update the weights and bias of a neural network in order to minimize the loss (error).

They work by adjusting the model parameters step-by-step using gradients so that predictions become more accurate.

Comparison of Gradient Descent, Adam, and RMSprop:

| Optimizer            | Working                                  | Advantages                                      | Disadvantages                       |
| -------------------- | ---------------------------------------- | ----------------------------------------------- | ----------------------------------- |
| **Gradient Descent** | Updates weights using full dataset       | Simple and easy to understand                   | Slow, may get stuck in local minima |
| **RMSprop**          | Uses moving average of squared gradients | Faster than basic GD, good for complex problems | Needs parameter tuning              |
| **Adam**             | Combines momentum + RMSprop ideas        | Very fast, most widely used                     | Slightly more complex               |

In Short:

Gradient Descent: Basic but slow

RMSprop: Better for faster convergence

Adam: Best overall (fast + efficient)



```python
#Use NumPy, Matplotlib, and Tensorflow/Keras for implementation.
#Question 6: Write a Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.

import numpy as np

# Input
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

# Weights and bias
weights = np.zeros(2)
bias = 0
lr = 0.1

# Training
for epoch in range(10):
    for i in range(len(X)):
        linear = np.dot(X[i], weights) + bias
        output = 1 if linear > 0 else 0
        error = y[i] - output

        weights += lr * error * X[i]
        bias += lr * error

print("Weights:", weights)
print("Bias:", bias)

# Testing
for i in X:
    linear = np.dot(i, weights) + bias
    print(i, "->", 1 if linear > 0 else 0)

```

```python
#Question 7: Implement and visualize Sigmoid, ReLU, and Tanh activation functions using Matplotlib.

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

sigmoid = 1/(1+np.exp(-x))
relu = np.maximum(0, x)
tanh = np.tanh(x)

plt.plot(x, sigmoid, label='Sigmoid')
plt.plot(x, relu, label='ReLU')
plt.plot(x, tanh, label='Tanh')

plt.legend()
plt.title("Activation Functions")
plt.show()
```

```python
#Question 8: Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.

import tensorflow as tf
from tensorflow import keras
import warnings
warnings.filterwarnings("ignore")

# Load data
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

X_train = X_train/255.0
X_test = X_test/255.0

# Model
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train
model.fit(X_train, y_train, epochs=5)

# Accuracy
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Accuracy:", test_acc)
```

```python
#Question 9: Visualize the loss and accuracy curves for a neural network model trained on the Fashion MNIST dataset. Interpret the training behavior.

import matplotlib.pyplot as plt
from tensorflow import keras

(X_train, y_train), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()

X_train = X_train/255.0
X_test = X_test/255.0

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(X_train, y_train, epochs=5)

# Plot
plt.plot(history.history['loss'], label='Loss')
plt.plot(history.history['accuracy'], label='Accuracy')
plt.legend()
plt.show()
```

```python
#Question 10: You are working on a project for a bank that wants to automatically detect
#fraudulent transactions. The dataset is large, imbalanced, and contains structured 
#features like transaction amount, merchant ID, and customer location. The goal is to
#classify each transaction as fraudulent or legitimate.
#Explain your real-time data science workflow:
#● How would you design a deep learning model (perceptron or multilayer NN)?
#● Which activation function and loss function would you use, and why?
#● How would you train and evaluate the model, considering class imbalance?
#● Which optimizer would be suitable, and how would you prevent overfitting?

import numpy as np
import pandas as pd
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Sample dummy dataset (replace with real data)
X = np.random.rand(1000, 5)
y = np.random.randint(0, 2, 1000)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Model
model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(5,)),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Class weights (handling imbalance)
class_weight = {0:1, 1:5}

# Train
history = model.fit(X_train, y_train,
                    epochs=10,
                    validation_split=0.2,
                    class_weight=class_weight)

# Evaluate
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Accuracy:", accuracy)

```
